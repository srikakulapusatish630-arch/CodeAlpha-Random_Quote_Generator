```python

import random
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output

quotes = [
    ("Believe you can and you're halfway there.", "Theodore Roosevelt"),
    ("Never give up on your dreams.", "Unknown"),
    ("Success is the sum of small efforts.", "Robert Collier"),
    ("Learning never exhausts the mind.", "Leonardo da Vinci"),
    ("Dream big and work hard.", "Unknown"),
    ("The future depends on what you do today.", "Mahatma Gandhi"),
    ("It always seems impossible until it's done.", "Nelson Mandela"),
    ("Your only limit is your mind.", "Unknown")
]

def show_random_quote(button=None):
    quote, author = random.choice(quotes)
    with output:
        clear_output(wait=True)
        display(HTML(f"""
        <div style="
            background:#f0f7ff;
            padding:25px;
            border-radius:12px;
            text-align:center;
            border-left:5px solid #2878d0;">
            <h3 style="color:#174a7e;">Random Quote</h3>
            <p style="font-size:20px;color:#222;">
                “{quote}”
            </p>
            <p style="font-size:16px;color:#555;">
                — {author}
            </p>
        </div>
        """))

title = widgets.HTML(
    "<h2 style='color:#2878d0'>Random Quote Generator</h2>"
)

new_button = widgets.Button(
    description="New Quote",
    button_style="primary",
    icon="refresh"
)

output = widgets.Output()

new_button.on_click(show_random_quote)

display(title, new_button, output)
show_random_quote()
```
